# Run the complete workflow

This notebook previews the pipeline and, when explicitly enabled, runs each stage in order. A full run is slow, so the saved results are not regenerated here.

## Setup

Run from the repository root after installing `.[notebooks,dev]`. The pipeline configuration keeps paths, data splits, random seeds, forecast horizons, and model settings together.

```python
import subprocess
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
CONFIG = PROJECT_ROOT / "configs" / "pipeline.toml"
assert CONFIG.exists(), CONFIG
```

## Preview the pipeline

```python
subprocess.run(
    [sys.executable, "scripts/run_pipeline.py", "--dry-run", "--config", str(CONFIG)],
    cwd=PROJECT_ROOT,
    check=True,
)
```

## Run the pipeline

Set the execution guard to `True` only when the raw dataset is available and a full run is intended.

```python
EXECUTE_PIPELINE = False

if EXECUTE_PIPELINE:
    subprocess.run(
        [sys.executable, "scripts/run_pipeline.py", "--execute", "--config", str(CONFIG)],
        cwd=PROJECT_ROOT,
        check=True,
    )
else:
    print("Execution disabled. Review the dry run above first.")
```
