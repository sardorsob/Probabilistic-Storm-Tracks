## Setup

Run from the repository root after installing `.[notebooks,dev]`. The
configuration file is the single source for paths, splits, seeds, horizons,
and model settings.

```python
from pathlib import Path
import subprocess
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
CONFIG = PROJECT_ROOT / "configs" / "pipeline.toml"
assert CONFIG.exists(), CONFIG
```

## Step 1: inspect the plan without running it

```python
subprocess.run(
    [sys.executable, "scripts/run_pipeline.py", "--dry-run", "--config", str(CONFIG)],
    cwd=PROJECT_ROOT,
    check=True,
)
```

## Step 2: opt into execution

Change the guard only when the raw dataset is present and a long full run is
intended. Execution delegates to the same runner shown in the dry run.

```python
EXECUTE_PIPELINE = False

if EXECUTE_PIPELINE:
    subprocess.run(
        [sys.executable, "scripts/run_pipeline.py", "--execute", "--config", str(CONFIG)],
        cwd=PROJECT_ROOT,
        check=True,
    )
else:
    print("Execution disabled. Review the dry run above first.")
```
